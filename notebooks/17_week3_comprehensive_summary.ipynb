### Part 2: Comprehensive Statistical Summary Table

**Objective:** Compile all p-values, hazard ratios, and effect sizes in one master table

**Contents:**
- All Cox model results
- All log-rank test results
- Model performance metrics
- Effect sizes with confidence intervals

```python
# Part 2: Comprehensive Statistical Summary
print("="*70)
print("PART 2: COMPREHENSIVE STATISTICAL SUMMARY TABLE")
print("="*70)

# Create master statistics table
all_statistics = []

print("\nCompiling all statistical results...")

# 1. Load survival analysis results
print("\n1. Loading survival analysis results...")
if (tables_dir / 'survival_analysis' / 'univariate_cox_hazard_ratios.csv').exists():
    cox_results = pd.read_csv(tables_dir / 'survival_analysis' / 'univariate_cox_hazard_ratios.csv')
    for _, row in cox_results.iterrows():
        all_statistics.append({
            'Analysis': 'Univariate Cox',
            'Feature': row['Variable'],
            'Metric': 'Hazard Ratio',
            'Value': row['HR'],
            'CI_Lower': row['CI_Lower'],
            'CI_Upper': row['CI_Upper'],
            'P_Value': row['P_Value'],
            'Significance': '***' if row['P_Value'] < 0.001 else ('**' if row['P_Value'] < 0.01 else ('*' if row['P_Value'] < 0.05 else 'NS'))
        })
    print(f"   Loaded {len(cox_results)} Cox regression results")

# 2. Load pathway survival results
print("2. Loading pathway survival results...")
if (tables_dir / 'pathway_survival' / 'pathway_survival_associations.csv').exists():
    pathway_results = pd.read_csv(tables_dir / 'pathway_survival' / 'pathway_survival_associations.csv')
    # Top 10 pathways
    for _, row in pathway_results.head(10).iterrows():
        all_statistics.append({
            'Analysis': 'Pathway Cox',
            'Feature': row['Pathway'],
            'Metric': 'Hazard Ratio',
            'Value': row['HR'],
            'CI_Lower': row['CI_Lower'],
            'CI_Upper': row['CI_Upper'],
            'P_Value': row['P_Value'],
            'Significance': '***' if row['FDR'] < 0.001 else ('**' if row['FDR'] < 0.01 else ('*' if row['FDR'] < 0.05 else 'NS'))
        })
    print(f"   Loaded top 10 pathway results (from {len(pathway_results)} total)")

# 3. Add ML model performance
print("3. Adding ML model performance...")
all_statistics.append({
    'Analysis': 'ML Model',
    'Feature': 'Random Survival Forest',
    'Metric': 'C-Index (Train)',
    'Value': 0.7754,
    'CI_Lower': np.nan,
    'CI_Upper': np.nan,
    'P_Value': np.nan,
    'Significance': 'N/A'
})
all_statistics.append({
    'Analysis': 'ML Model',
    'Feature': 'Random Survival Forest',
    'Metric': 'C-Index (Validation)',
    'Value': 0.6959,
    'CI_Lower': np.nan,
    'CI_Upper': np.nan,
    'P_Value': np.nan,
    'Significance': 'N/A'
})
all_statistics.append({
    'Analysis': 'ML Model',
    'Feature': 'Random Survival Forest',
    'Metric': 'C-Index (Test)',
    'Value': 0.7164,
    'CI_Lower': np.nan,
    'CI_Upper': np.nan,
    'P_Value': np.nan,
    'Significance': 'N/A'
})

# Create master DataFrame
stats_df = pd.DataFrame(all_statistics)

# Save master statistics table
stats_path = summary_dir / 'comprehensive_statistics_summary.csv'
stats_df.to_csv(stats_path, index=False)
print(f"\n✅ Saved: {stats_path}")

# Display summary
print("\n" + "="*70)
print("STATISTICAL SUMMARY")
print("="*70)

print(f"\nTotal statistical results compiled: {len(stats_df)}")
print(f"\nBy analysis type:")
print(stats_df.groupby('Analysis').size())

print(f"\nBy significance level:")
print(stats_df['Significance'].value_counts())

# Key findings table
print("\n" + "="*70)
print("KEY STATISTICAL FINDINGS (Top 10)")
print("="*70)

significant_results = stats_df[stats_df['Significance'].isin(['***', '**', '*'])].sort_values('P_Value')
print(significant_results[['Analysis', 'Feature', 'Metric', 'Value', 'P_Value', 'Significance']].head(10).to_string(index=False))

print("\n✅ Part 2 Complete: Comprehensive statistics compiled")
```
